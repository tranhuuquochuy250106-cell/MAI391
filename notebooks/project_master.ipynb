```python
from pathlib import Path
import os
import subprocess
import sys

here = Path.cwd()
root = here if (here / "run_q1.py").exists() else here.parent
if not (root / "run_q1.py").is_file():
    raise FileNotFoundError("Mở notebook từ thư mục MAI391_Final hoặc notebooks.")

def run_question(number):
    env = dict(os.environ, PYTHONIOENCODING="utf-8", MPLBACKEND="Agg")
    result = subprocess.run(
        [sys.executable, str(root / f"run_q{number}.py")],
        cwd=root, env=env, capture_output=True, encoding="utf-8", errors="replace",
    )
    print(result.stdout, end="")
    if result.stderr:
        print(result.stderr, end="")
    result.check_returncode()
```
